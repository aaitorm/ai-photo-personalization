```python
import pathlib 
NBS_DIR = pathlib.Path().resolve()
REPO_DIR = NBS_DIR.parent
DATA_DIR = REPO_DIR / "data"
GENERATED_DIR = DATA_DIR / "generated"
GENERATED_DIR.mkdir(exist_ok=True, parents=True)
REFERENCES_DIR = DATA_DIR / "references"
REFERENCES_DIR.mkdir(exist_ok=True, parents=True)
```

```python
import replicate    
from decouple import config 
from replicate.client import Client

REPLICATE_API_TOKEN=config("REPLICATE_API_TOKEN")
REPLICATE_MODEL= config("REPLICATE_MODEL", default="aaitorm/superme-aitor-v1")
REPLICATE_MODEL_VERSION= config("REPLICATE_MODEL_VERSION", default="76fd9d004a50354c0d509dda96b7945139f83c9bcd86dc011010ffc11c3d0ea7")

replicate_client= Client(api_token=REPLICATE_API_TOKEN)
```

```python
from PIL import Image

from PIL import Image
path=REFERENCES_DIR / "spider-man.webp"
img = Image.open(path)
s=REFERENCES_DIR / "spider-man.jpg"
img.save(s, "JPEG")
```

```python
# reference_image_path = REFERENCES_DIR / "spider-man.webp"
# image = open(reference_image_path, 'rb')
reference_image_url = (
    "https://github.com/aaitorm/super-me-photo-ai-api/blob/start/"
    "data/references/spider-man.jpg?raw=1    "
)
print(reference_image_url)
```

```python
model = f"{REPLICATE_MODEL}:{REPLICATE_MODEL_VERSION}"
prompt = "a photo of TOK adult man dressed up for a professional photo shoot"
responses = replicate_client.run(
    model,
    input={
        "image": reference_image_url,
        "model": "dev",
        "prompt": prompt  ,
        "output_format": "jpg",
        "num_outputs": 4, 
    }
)
#print(responses)

# # To access the file URL:
# print(responses[0].url)
# #=> "http://example.com"

# # To write the file to disk:
# with open("my-image.png", "wb") as file:
#     file.write(output[0].read())
```

```python
import random
session_id = random.randint(1_000, 40_000)
for i, output in enumerate(responses):
    fname = f"{i}-{session_id}.jpg"
    outpath = GENERATED_DIR / fname
    with open(outpath, 'wb') as f:
        f.write(output.read())
```
